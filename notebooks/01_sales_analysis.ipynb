# Nova Insights — Sales Analysis

This notebook mirrors CodeNova Task 1 using the bundled synthetic transaction data. All figures and conclusions should be interpreted as a workflow demonstration, not real-company results.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / "data").exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
from analytics import load_sales, clean_sales
import pandas as pd
import matplotlib.pyplot as plt
raw = load_sales(ROOT / "data" / "sales_raw.csv")
sales, audit = clean_sales(raw)
sales.head()
```

```python
revenue = sales.revenue.sum()
profit = sales.profit.sum()
orders = sales.order_id.nunique()
print({"revenue": round(revenue, 2), "profit": round(profit, 2), "margin_pct": round(profit/revenue*100, 2), "AOV": round(revenue/orders, 2), "units": int(sales.quantity.sum())})
```

```python
category_summary = sales.groupby("category", as_index=False).agg(revenue=("revenue", "sum"), profit=("profit", "sum"), units=("quantity", "sum")).sort_values("revenue", ascending=False)
category_summary
```

```python
category_summary.plot(kind="bar", x="category", y="revenue", legend=False, figsize=(9, 4), title="Revenue by category")
plt.ylabel("Revenue (₹)"); plt.xticks(rotation=20, ha="right"); plt.tight_layout(); plt.show()
```
